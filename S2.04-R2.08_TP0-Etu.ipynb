**13.** Créer un `numpy.array` `X` rempli de "1", ayant 3 colonnes et `N` lignes (sur l'Aide-Mémoire, partie "Création de tableaux").

```python
X = np.zeros([N,3])
X=X+1
print(X)
```

<div class="alert alert-info" role="alert">

**14.** Changer la 2e colonne de `X` en y mettant le contenu du tableau `P_TB`.
Puis changer la 3e colonne de `X` pour y mettre le taux d'admission de chaque formation : c'est-à-dire le nombre d'admis.es divisé par le nombre de candidat.es.

```python
X[:,1]=P_TB
X[:,2]=(Adm)/(arr_lannion[:,1])
print(X)
```

<div class="alert alert-info" role="alert">


**15.** Créer `tX`, la matrice transposée $^t X$ de `X`. Puis calculer le produit matriciel $^tX X$ (dans l'Aide-Mémoire, se servir de la partie "Opérations mathématiques").

```python
tX=np.transpose(X)

XtX=tX @ X
print(XtX)
```

<div class="alert alert-info" role="alert">


**16.** Calculer l'inverse $ (^t X X )^{-1}$ du produit matriciel calculé ci-dessus (dans l'Aide-Mémoire, voir la partie "`numpy.linalg`: méthodes d'algèbre linéaire").

```python
import numpy.linalg as la

print(la.inv(XtX))
```
